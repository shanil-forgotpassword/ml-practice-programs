```python
from pydoc import classify_class_attrs

import numpy as  np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

from pandas.core._numba import kernels

warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42)
from sklearn.neighbors import KNeighborsClassifier
model_knn = KNeighborsClassifier()
```

```python
model_knn.fit(X_train, y_train)
```

```python
model_knn.score(X_test, y_test)   #overfit
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma='auto')
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)#overfit
```

```python
#now let's use the grid search CV
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV(model_svm,{ 'C' : [1,10,20,30], 'kernel':['rbf', 'linear']},cv = 5 , return_train_score= False )
```

```python
classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
param_grid = {
    'n_neighbors': list(range(1, 14)),
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

classifier2 = GridSearchCV(
    estimator=model_knn,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)
classifier2.fit(X,y)
```

```python
classifier2.cv_results_
```

```python
results2 = pd.DataFrame(classifier2.cv_results_)
```

```python
results2
```

```python
results2[['param_n_neighbors','param_metric','param_weights','mean_test_score']]
```
