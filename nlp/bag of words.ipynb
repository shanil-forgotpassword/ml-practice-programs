**countvectorizer convert a collection of text documents to a matrix of token counts. this gives a sparse representation.**

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love pizza",
    "Pizza is best",
    "i love pasta",
    "pasta is great"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

print("vocabulary:", vectorizer.get_feature_names_out())
print("\n BoW Matrix:\n", X.toarray())
```

**diffrence with ngrams**

```python
from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love pizza",
    "Pizza is best",
    "i love pasta",
    "pasta is great"
]

vectorizer = CountVectorizer(ngram_range=(1, 2))
X = vectorizer.fit_transform(documents)

print("vocabulary:", vectorizer.get_feature_names_out())
print("\n BoW Matrix:\n", X.toarray())
```
